```python
import pandas as pd
import numpy as np
import os

# Load raw data
df = pd.read_csv('../data/hour.csv')
print('Raw shape:', df.shape)

# Basic cleaning
df = df.drop(['instant', 'casual', 'registered'], axis=1, errors='ignore')
df['dteday'] = pd.to_datetime(df['dteday'])

# Feature engineering
df['year'] = df['dteday'].dt.year
df['month'] = df['dteday'].dt.month
df['day'] = df['dteday'].dt.day
df['hour'] = df['hr']
df['is_weekend'] = df['weekday'].isin([0, 6]).astype(int)
df['is_morning_rush'] = df['hr'].between(7, 9).astype(int)
df['is_evening_rush'] = df['hr'].between(16, 19).astype(int)
df['temp_hum'] = df['temp'] * df['hum']

# Drop original hour (we have engineered version)
df = df.drop('hr', axis=1)

# Save engineered data WITH dteday (needed for split)
os.makedirs('../data/processed', exist_ok=True)
df.to_csv('../data/processed/engineered_bike_data.csv', index=False)
print('Engineered data saved!')
df.head()
```
